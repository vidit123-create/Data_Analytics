```
from great_expectations.dataset import SparkDFDataset
from pyspark.sql import SparkSession
from pyspark.sql import Window as w
from pyspark.sql import functions as f

spark = SparkSession.builder.appName('Program2').getOrCreate()
df = spark.read.csv('/FileStore/data_preprocessing_dataset.csv',header=True,inferSchema=True)
df.display()
```

```
# Detection of missing values in all the columns :

my_list, list_1 = df.columns,[]
list_2 = ['cust_id_d','age_d','gender_d','income_d','spending_d','membership_d','city_d','date_d']
for i, j in zip(my_list,list_2):
   # column_name = input("Column Name : ")
   df = df.withColumn(j,f.col(i).isNull())
   missing_value_count = df.filter(df[j] == True).select(f.count(j)).collect()[0][0]
   list_1.append(missing_value_count)
print(f"Missing Values Count in all the columns : {list_1}")
```

```
# Total number of rows in Pyspark DataFrame:

total_no_of_rows = df.count()
print(f"Total number of rows : {total_no_of_rows}")
```

```
# Percentage of missing/null values in each column:

for i,j in zip(my_list,list_1):
    print(f"Percent of missing value in {i} : {(j/total_no_of_rows)*100}")
```

```
# Based on the percentage of missing values we have three columns i.e. Age, Gender and Annual_Income having less than 30% missing values hence will use dropna() 
```

```
df = df.dropna(subset=['Age','Gender','Annual_Income'],how='any')
df.show()
```

```
# Re-Validation of Missing values in all the columns :
my_list, list_1 = df.columns,[]
list_2 = ['cust_id_d','age_d','gender_d','income_d','spending_d','membership_d','city_d','date_d']
for i, j in zip(my_list,list_2):
   # column_name = input("Column Name : ")
   df = df.withColumn(j,f.col(i).isNull())
   missing_value_count = df.filter(df[j] == True).select(f.count(j)).collect()[0][0]
   list_1.append(missing_value_count)
print(f"Missing Values Count in all the columns : {list_1}")
```

```
for i in list_2:
    df = df.drop(i)
df1 = df
df1.display()
```
