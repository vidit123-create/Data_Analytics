```
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
spark = SparkSession.builder.appName('Program1').getOrCreate()
df = spark.read.csv('/FileStore/skewed_sales_data_10000.csv',header=True,inferSchema=True)
df.display()
```

```
# Detection of data skewness based on display() with shows visualization using histogram:
df.select('Sale_Amount').display()
# Total count of values in Sale_Amount
count_before_outliers = df.agg(f.count('Sale_Amount')).collect()[0][0]
print(f"Count : {count_before_outliers}")
```

```
"""
Since the column contains skewed data and according to the graph it is right skewed in nature which means 
most of the points are on the left and right there is a long tail which also stats that median < mean
"""
# Calculation of Mean and Median
average_value = df.agg(f.avg('Sale_Amount').alias('Average_Value')).collect()[0][0]
print(f"Mean of the Sale_Amount : {round(average_value,2)}")

from pyspark.sql import Window as w
middle_index = (df.agg(f.count('Sale_Amount')).collect()[0][0])//2
rule = w.orderBy(df['Sale_Amount'])
df = df.withColumn('rank_sale_amount',f.row_number().over(rule))
middle_value= df.filter(df['rank_sale_amount'] == middle_index).select('Sale_Amount').drop('rank_sale_amount').collect()[0][0]
print(f"Median of the Sale Amount : {round(middle_value,2)}")
```

```
"""
As we have checked that median is less than mean and data is right skewed so it contains Outliers. To make the column appropriate will perform Outlier Detection using IQR Method :
"""
# Calculating 25th, 50th and 75th Percentile -
q1,q2,q3 = df.approxQuantile('Sale_Amount',[0.25,0.50,0.75],0.0)
print(f"25th Percentile : {q1}"+'\n'+f"50th Percentile : {q2}"+'\n'+f"75th Percentile : {q3}")

# Calculating Inter Quartile Range :
IQR = q3 - q1
print(f"IQR : {round(IQR,2)}")

# Calculating Lower extreme and Upper extreme value :
Lx,Ux = q1-1.5*IQR, q3+1.5*IQR
print(f"Lower Extreme Value : {round(Lx,2)}"+'\n'+f"Upper Extreme Value : {round(Ux,2)}")
```

```
# Filter the value based on Lx and Ux :
df1 = df.filter((df['Sale_Amount']>Lx) & (df['Sale_Amount']<Ux)).select('Sale_Amount')
df1.display()
count_after_outliers = df1.agg(f.count('Sale_Amount')).collect()[0][0]
print(f"Count : {count_after_outliers}")
```

```
# Outlier data points in Sale Amount :
total_outlier_in_sale_amount = count_before_outliers - count_after_outliers
print(f"Outliers In Sale Amount : {total_outlier_in_sale_amount}")
```

```
# Re-Validation of Sale Amount for Data Skewness :
df1.select('Sale_Amount').display()
```

```
# Calculation of Mean and Median
average_value = df1.agg(f.avg('Sale_Amount').alias('Average_Value')).collect()[0][0]
print(f"Mean of the Sale_Amount : {round(average_value,2)}")

from pyspark.sql import Window as w
middle_index = (df1.agg(f.count('Sale_Amount')).collect()[0][0])//2
rule = w.orderBy(df1['Sale_Amount'])
df1 = df1.withColumn('rank_sale_amount',f.row_number().over(rule))
middle_value= df1.filter(df1['rank_sale_amount'] == middle_index).select('Sale_Amount').drop('rank_sale_amount').collect()[0][0]
print(f"Median of the Sale Amount : {round(middle_value,2)}")
```

```
# Before Outlier detection the difference between mean and median was around 90.1 and after outlier detection the difference is 57.5 which means that to an approximate value data skewness has been decreased in Sale Amount column.
```
